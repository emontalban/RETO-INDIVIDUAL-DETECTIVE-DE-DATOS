# 2. ¿Qué resultado te ha sorprendido?

Me ha sorprendido que una columna aparentemente sencilla como Type necesitara varias decisiones de preparación. Algunos hechizos tienen varios tipos, hay espacios especiales y existen diferencias de mayúsculas. Antes de contar categorías, hemos tenido que entender cómo estaban registradas.

También me ha llamado la atención que House incluya otras escuelas de magia, además de las cuatro casas de Hogwarts. El nombre de una columna no siempre describe con precisión todo su contenido.

Hemos encontrado patrones de distribución, aunque todavía no hemos analizado una evolución en el tiempo:
- Hay más personajes masculinos que femeninos registrados en tres de las cuatro casas.
- Human es la categoría predominante entre los personajes con casa registrada: 87 de 97. Las etiquetas humanas con características adicionales se conservaron por separado.

# 4. ¿Qué decisión podría tomar una empresa u organización?
Una organización que prepare una exposición, una actividad educativa o un catálogo sobre Harry Potter podría utilizar estos resultados para organizar sus contenidos:
- Comprobar la representación de las casas y buscar más información sobre las que tienen menos personajes documentados.
- Crear secciones de hechizos según sus tipos, indicando que algunos pertenecen a varias categorías.
- Priorizar la documentación de la dificultad de las pociones, porque es un dato ausente en gran parte de la tabla.
Por ejemplo, antes de diseñar actividades sobre pociones para principiantes, convendría completar y verificar las dificultades, en lugar de seleccionar únicamente entre los pocos registros disponibles.

No utilizaríamos estos conteos para decidir qué productos vender más: faltan datos de preferencias, demanda o ventas.
