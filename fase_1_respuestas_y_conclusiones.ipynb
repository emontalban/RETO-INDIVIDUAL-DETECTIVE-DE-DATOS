## 3. ¿Hay datos incompletos o problemas de calidad?

### Tipos de datos

En personajes, `Id` es `int64`; las otras 14 columnas son `object`.   
Las cinco columnas de pociones y las cinco de hechizos son `object`.

`Id` es un identificador: su media no aporta información útil. `Birth` y `Death` están almacenadas como texto y presentan fechas de formatos distintos; necesitarían revisión antes de realizar cálculos temporales.

### Valores ausentes en hechizos

Los nombres, tipos y efectos están completos según `isna()`.   
La luz no está registrada en 155 hechizos y la fórmula verbal falta en 7.   
La ausencia de una fórmula o de luz en la tabla no prueba que no exista.

### Duplicados y decisiones de preparación

Las tres tablas devuelven cero filas completamente duplicadas. Esta comprobación no descarta nombres repetidos con otros datos distintos.

Conservamos las filas y los valores ausentes. No rellenamos información sin evidencia y no eliminamos registros solo por estar incompletos. Antes de contar categorías revisaremos su escritura y posibles espacios adicionales.

`isna()` detecta valores que pandas reconoce como ausentes; no detecta por sí solo todos los problemas de calidad, como etiquetas inconsistentes o textos compuestos únicamente por espacios.

La variable que me parece más interesante es `House`, porque permite comparar la cantidad de personajes registrados en cada casa y explorar su distribución según `Gender` o `Blood status`.
