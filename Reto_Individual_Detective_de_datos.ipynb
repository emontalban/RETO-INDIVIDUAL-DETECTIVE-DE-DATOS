## Fase 1. Conoce tus datos

### 1. Importar las librerías necesarias

Importamos pandas para cargar y analizar los datos, matplotlib.pyplot
para crear gráficos y kagglehub para obtener los archivos desde Kaggle. O puedes descargar el dataset de la pagina de kaggle https://www.kaggle.com/datasets/gulsahdemiryurek/harry-potter-dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
import kagglehub
```

#### Obtener el dataset desde Kaggle

Descargamos los archivos del dataset en el entorno de Google Colab.

Guardamos en la variable path la ruta de la carpeta donde se encuentran.
Este paso obtiene los archivos, pero todavía no los carga en un DataFrame.  
Esto significa:  
- kagglehub.dataset_download(...): descarga los archivos del dataset, o utiliza una copia si ya está disponible en la caché.
- "gulsahdemiryurek/harry-potter-dataset": identifica el dataset mediante el usuario que lo publicó y su nombre.
- path =: guarda la ruta de la carpeta en una variable llamada path. Guarda una ubicación, no una tabla de datos.

```python
# Descargamos el dataset y guardamos la ruta de su carpeta.
path = kagglehub.dataset_download(
    "gulsahdemiryurek/harry-potter-dataset"
)
```

Antes de cargar los datos, mostramos los archivos disponibles para
identificar el archivo de personajes y conocer su ubicación exacta.

```python
# importamos la libreria para trabajar con archivos
from pathlib import Path

# Buscamos todos los elementos de la carpeta y sus subcarpetas.
# El asterisco "*" permite buscar cualquier nombre.
for archivo in Path(path).rglob("*"):
    # Comprobamos que el elemento sea un archivo.
    if archivo.is_file():
        print(archivo.relative_to(path))
```

### 2. Cargar el archivo CSV en un DataFrame

Seleccionamos Characters.csv, el archivo de personajes del dataset.

Utilizamos pandas para leerlo y guardamos su contenido en un DataFrame
llamado df. Un DataFrame es una tabla organizada en filas y columnas.

Este archivo utiliza el punto y coma como separador, por lo que lo
indicamos mediante el parámetro sep=";".

```python
# Construimos la ruta de los arvhivos.
ruta_personajes = Path(path) / "Characters.csv"
ruta_pociones = Path(path) / "Potions.csv"
ruta_hechizos = Path(path) / "Spells.csv"

# Leemos el archivo y guardamos la tabla en df.
df_personajes = pd.read_csv(ruta_personajes, sep=";")
df_pociones = pd.read_csv(ruta_pociones, sep=";")
df_hechizos = pd.read_csv(ruta_hechizos, sep=";")
```

### ¿Cuántos registros y variables tiene?

La tabla de personajes contiene 140 registros y 15 variables.
Cada fila representa un personaje.

La tabla de pociones contiene 72 registros y 5 variables.
Cada fila representa una poción.

### ¿Hay datos incompletos o problemas de calidad?

Sí, ambas tablas presentan valores ausentes.

La ausencia de información no implica necesariamente un error.
