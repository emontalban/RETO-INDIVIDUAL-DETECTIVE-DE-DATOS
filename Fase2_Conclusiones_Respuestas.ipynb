## Pregunta 1. ¿Qué casa de Hogwarts tiene más personajes registrados?

Gryffindor es la casa con más personajes registrados, con 38.
Le siguen Slytherin, con 28; Ravenclaw, con 18; y Hufflepuff, con 13.

La diferencia entre Gryffindor y Hufflepuff es de 25 personajes.

Para esta comparación seleccionamos únicamente las cuatro casas de
Hogwarts. Excluimos los valores ausentes y las otras escuelas que
también aparecen en la columna House.

El análisis incluye 97 personajes con una casa de Hogwarts registrada.
Los resultados describen la composición del dataset y no permiten
afirmar que una casa tenga más estudiantes en todo el universo
de Harry Potter.

## Pregunta 3. ¿Qué tipos de hechizos aparecen con mayor frecuencia?


Después de limpiar los espacios, unificar las mayúsculas y separar
los tipos indicados mediante comas, las categorías más frecuentes son:

| Tipo | Apariciones |
|---|---:|
| charm | 160 |
| jinx | 33 |
| curse | 27 |
| transfiguration | 23 |
| hex | 22 |

Charm es la categoría más frecuente y supera ampliamente a las siguientes.

Un hechizo puede tener varios tipos registrados. Por eso, los conteos
no representan grupos independientes y su suma puede superar los
301 hechizos del dataset.
Un hechizo puede pertenecer a varios tipos. Para tenerlo en cuenta,
separamos las categorías indicadas mediante comas y contamos cada
una por separado.

Así, un hechizo clasificado como Charm, Conjuration contribuye
al conteo de charm y al de conjuration. Por este motivo, la suma
de las apariciones puede superar el número total de hechizos.

Las etiquetas sin comas se conservan como categorías independientes.

## Pregunta 4 ¿Qué especies son más comunes entre los personajes con una casa de Hogwarts registrada?

La categoría Human es la más frecuente, con 87 de los 97 personajes
analizados, aproximadamente el 89,7 %.

La segunda categoría más frecuente es Ghost, con 5 personajes.
Las otras cinco categorías tienen un personaje cada una:

- Half-Human/Half-Giant.
- Werewolf.
- Human (Werewolf traits).
- Human(goblin ancestry).
- Human (Metamorphmagus).

Antes del conteo, limpiamos los espacios especiales y los espacios
al principio y al final de las etiquetas.

Conservamos como categorías independientes las etiquetas que incluyen
ascendencia o características adicionales. Por tanto, el 89,7 %
corresponde exclusivamente a la etiqueta Human.

La columna Species mezcla especies con condiciones o características
de los personajes. Los resultados reflejan esa clasificación del archivo.

Este análisis incluye únicamente a los personajes con una de las
cuatro casas de Hogwarts registrada; no representa a todos los
personajes relacionados con la escuela.

## Pregunta 5 ¿Qué nivel de dificultad es más frecuente entre las pociones que tienen ese dato registrado?

Advanced es la categoría más frecuente, con 14 pociones, seguida
de Beginner, con 6. Las otras cinco categorías tienen una poción
cada una.

Entre las 25 pociones con dificultad registrada:

- El 56 % pertenece a Advanced.
- El 24 % pertenece a Beginner.
- El 20 % restante se reparte entre las otras cinco categorías.

Antes del conteo, sustituimos los espacios especiales por espacios
normales y eliminamos los espacios de los extremos.

Conservamos las etiquetas completas, incluidos los rangos como
Simple to moderate y Moderate to advanced.

Estos resultados describen únicamente las pociones con dificultad
registrada. Como falta este dato en 47 de las 72 pociones, no podemos
concluir que la mayoría de todas las pociones sean de nivel avanzado.
